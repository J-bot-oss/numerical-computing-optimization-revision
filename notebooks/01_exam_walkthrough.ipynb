## Week 1 — Residuals and MSE
For actual `[5,7,9]` and predictions `[4,8,7]`, predict residuals and MSE before running the cell.


```python
import numpy as np
actual = np.array([5,7,9], dtype=float)
predicted = np.array([4,8,7], dtype=float)
residuals = actual - predicted
mse = np.mean(residuals**2)
print('residuals:', residuals, 'MSE:', mse)
```

## Week 2 — Bisection sign checking
Change the function and starting bracket only after verifying the IVT assumptions.


```python
f = lambda x: x**3 - x - 2
a, b = 1.0, 2.0
assert f(a)*f(b) < 0
for n in range(1, 4):
    c = (a+b)/2
    fc = f(c)
    print(n, (a,b), c, fc)
    if fc == 0: break
    if f(a)*fc < 0: b = c
    else: a = c
print('Final bracket:', (a,b))
```

## Week 4 — Constrained optimization
Explain why the unconstrained minimum may violate the feasible range.


```python
J = lambda w: (w-3)**2
grid = np.linspace(0, 2, 201)
w_constrained = grid[np.argmin(J(grid))]
print('Unconstrained minimum at w=3, loss=', J(3))
print('Constrained candidate:', w_constrained, 'loss=', J(w_constrained))
```
